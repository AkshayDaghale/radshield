```python
# 2. Add current directory to Python Path & Define Self-Contained KAVACH Pipeline
import os, sys, math, joblib, torch
import torch.nn as nn
import numpy as np
import pandas as pd
from sklearn.preprocessing import StandardScaler

print("PyTorch version:", torch.__version__)
print("GPU available:", torch.cuda.is_available())
```

```python
# 3. Define Synthetic Space Weather Data Generator & Physics Features
def generate_space_weather_data(days=14):
    n = days * 288
    date_range = pd.date_range("2024-01-01", periods=n, freq="5min")
    np.random.seed(42)
    vsw = 400 + 100 * np.sin(np.linspace(0, 4*np.pi, n)) + np.random.normal(0, 15, n)
    bz = 2 * np.cos(np.linspace(0, 6*np.pi, n)) + np.random.normal(0, 2, n)
    by = 3 * np.sin(np.linspace(0, 5*np.pi, n)) + np.random.normal(0, 2, n)
    np_density = 5 + 3 * np.cos(np.linspace(0, 8*np.pi, n)) + np.random.exponential(1, n)
    kp = np.clip(2 + 1.5 * np.sin(np.linspace(0, 4*np.pi, n))**2 + np.random.normal(0, 0.3, n), 0, 9)
    dst = -10 - 20 * (kp / 3.0)**1.5 + np.random.normal(0, 3, n)
    ae = 100 + 150 * (kp / 2.0) + np.random.exponential(50, n)
    ulf_power = -3.5 + 0.5 * (kp / 3.0) + np.random.normal(0, 0.2, n)
    
    log_flux = pd.Series(2.3 + 0.005 * (vsw - 400) + 0.3 * (kp - 2) + 0.4 * (ulf_power + 3.5)).ewm(span=18).mean().values
    flux = np.clip(10 ** log_flux, 0.1, None)
    
    df = pd.DataFrame({
        "flux": flux, "log_flux": np.log10(flux), "Vsw": vsw, "BZ_GSM": bz, "BY_GSM": by,
        "BT": np.sqrt(by**2 + bz**2), "Np": np_density, "KP": kp, "DST": dst, "AE": ae, "ULF_power": ulf_power
    }, index=date_range)
    
    for lag_steps, label in [(12, "1h"), (36, "3h"), (72, "6h"), (144, "12h"), (288, "24h")]:
        df[f"flux_lag_{label}"] = df["log_flux"].shift(lag_steps)
    
    Bt = df["BT"]
    theta = np.arctan2(df["BY_GSM"], df["BZ_GSM"])
    df["Ec"] = np.clip((df["Vsw"]**(4/3)) * ((Bt * np.abs(np.sin(theta/2)))**(8/3)), 0, None)
    df["Pdyn"] = np.clip(0.5 * 1.67e-27 * (df["Np"]*1e6) * ((df["Vsw"]*1e3)**2) * 1e9, 0.1, 50.0)
    df["Bz_neg"] = (df["BZ_GSM"] < 0).astype(int)
    df["Bz_neg_dur"] = df["Bz_neg"].groupby((df["Bz_neg"] != df["Bz_neg"].shift()).cumsum()).cumcount() * 5.0
    df["dDst_dt"] = df["DST"].diff() / 5.0
    df["AE_1h"] = df["AE"].rolling(12, min_periods=1).mean()
    df["regime"] = 0
    return df.bfill().fillna(0)

print("Data generator ready.")
```

```python
# 4. Define PyTorch Temporal Fusion Transformer (TFT) Architecture
class GatedResidualNetwork(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim, dropout=0.1):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.fc2 = nn.Linear(hidden_dim, output_dim)
        self.dropout = nn.Dropout(dropout)
        self.gate = nn.Linear(input_dim, output_dim)
        self.layer_norm = nn.LayerNorm(output_dim)
        self.skip = nn.Linear(input_dim, output_dim) if input_dim != output_dim else nn.Identity()
    def forward(self, x):
        h = self.dropout(self.fc2(torch.relu(self.fc1(x))))
        g = torch.sigmoid(self.gate(x))
        return self.layer_norm(self.skip(x) + g * h)

class KAVACH_TFT(nn.Module):
    def __init__(self, num_features=19, hidden_size=128, lstm_layers=2, num_quantiles=5):
        super().__init__()
        self.vsn = nn.ModuleList([GatedResidualNetwork(1, hidden_size, hidden_size) for _ in range(num_features)])
        self.vsn_weights = nn.Linear(num_features, num_features)
        self.lstm = nn.LSTM(hidden_size, hidden_size, num_layers=lstm_layers, batch_first=True)
        self.attention = nn.MultiheadAttention(embed_dim=hidden_size, num_heads=4, batch_first=True)
        self.grn_dec = GatedResidualNetwork(hidden_size, hidden_size, hidden_size)
        self.quantile_head = nn.Linear(hidden_size, 144 * num_quantiles)
    def forward(self, x):
        b, seq_len, n_feat = x.shape
        vsn_outputs = [self.vsn[i](x[:, :, i:i+1]) for i in range(min(n_feat, 19))]
        vsn_stack = torch.stack(vsn_outputs, dim=-1)
        attn_scores = torch.softmax(self.vsn_weights(x.mean(dim=1)), dim=-1).unsqueeze(1).unsqueeze(2)
        vsn_fused = (vsn_stack * attn_scores).sum(dim=-1)
        lstm_out, _ = self.lstm(vsn_fused)
        attn_out, _ = self.attention(lstm_out, lstm_out, lstm_out)
        dec_out = self.grn_dec(attn_out[:, -1, :])
        return self.quantile_head(dec_out).view(b, 144, 5)

class PinballLoss(nn.Module):
    def __init__(self, quantiles=[0.10, 0.25, 0.50, 0.75, 0.90]):
        super().__init__()
        self.quantiles = quantiles
    def forward(self, pred, target):
        loss = 0.0
        for i, q in enumerate(self.quantiles):
            errors = target - pred[:, :, i]
            loss += torch.max((q - 1) * errors, q * errors).mean()
        return loss

print("PyTorch TFT architecture loaded.")
```

```python
# 5. Train Model on GPU (Use GPU T4 x2 on Kaggle, not P100)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# Check compute capability to avoid Tesla P100 (sm_60) incompatibility with PyTorch 2.4+
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    cap = torch.cuda.get_device_capability(0)
    print(f"Detected GPU: {gpu_name} (CUDA Capability {cap[0]}.{cap[1]})")
    if cap[0] < 7:
        print("⚠️ Warning: Tesla P100 (sm_60) is not supported by PyTorch 2.4+.")
        print("👉 Please switch Kaggle Session Options -> Accelerator to: GPU T4 x2")
        device = torch.device("cpu")
print("Training device in use:", device)

data = generate_space_weather_data(days=14)
feature_cols = [
    "log_flux", "flux_lag_1h", "flux_lag_3h", "flux_lag_6h", "flux_lag_12h", "flux_lag_24h",
    "Vsw", "BZ_GSM", "BY_GSM", "Np", "Pdyn", "Ec", "DST", "dDst_dt", "KP", "AE_1h",
    "ULF_power", "Bz_neg_dur", "regime"
]
scaler = StandardScaler()
scaled_data = scaler.fit_transform(data[feature_cols].values)

seq_len, pred_len = 288, 144
X_samples, y_samples = [], []
for i in range(len(scaled_data) - seq_len - pred_len):
    X_samples.append(scaled_data[i : i + seq_len])
    y_samples.append(data["log_flux"].values[i + seq_len : i + seq_len + pred_len])

X_tensor = torch.tensor(np.array(X_samples), dtype=torch.float32).to(device)
y_tensor = torch.tensor(np.array(y_samples), dtype=torch.float32).to(device)

model = KAVACH_TFT().to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
criterion = PinballLoss()

epochs = 10
batch_size = 32
n_samples = len(X_tensor)

print(f"Starting training on {n_samples} samples...")
model.train()
for epoch in range(epochs):
    epoch_loss = 0.0
    indices = torch.randperm(n_samples)
    for b in range(0, n_samples, batch_size):
        batch_idx = indices[b : b + batch_size]
        bx, by = X_tensor[batch_idx], y_tensor[batch_idx]
        optimizer.zero_grad()
        q_preds = model(bx)
        loss = criterion(q_preds, by)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
    print(f"Epoch {epoch+1}/{epochs} - Loss: {epoch_loss / (n_samples/batch_size):.4f}")

# Save model checkpoints to Kaggle Output
torch.save(model.state_dict(), "/kaggle/working/kavach_tft_v1.pt")
joblib.dump(scaler, "/kaggle/working/scaler.pkl")
print("✅ Training complete! Checkpoints saved to /kaggle/working/kavach_tft_v1.pt and /kaggle/working/scaler.pkl")
```

```python
# 6. MLOps Cloud Bridge: Automatically push trained weights to Hugging Face Model Hub
from huggingface_hub import HfApi
import os

hf_token = os.environ.get("HF_TOKEN")
repo_id = "DigiIndia/kavach-weights"

if hf_token:
    api = HfApi()
    api.upload_file(path_or_fileobj="/kaggle/working/kavach_tft_v1.pt", path_in_repo="kavach_tft_v1.pt", repo_id=repo_id, token=hf_token)
    api.upload_file(path_or_fileobj="/kaggle/working/scaler.pkl", path_in_repo="scaler.pkl", repo_id=repo_id, token=hf_token)
    print(f"Successfully published new GPU model weights to Hugging Face Hub: {repo_id}")
else:
    print("Note: Set HF_TOKEN secret in Kaggle to enable zero-click auto-publishing to Cloud Hub.")
```
